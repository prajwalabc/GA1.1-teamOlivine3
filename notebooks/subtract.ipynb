### Subtract

This function will subtract number b from number a. It will also check that both b and a are numbers

```python
def subtract(a, b):
    type(a)
    assert type(a) == int or float, "A should be a number"
    assert type(b) == int or float, "b should be a number"
    return a-b

subtract(10, 5)
assert subtract(3, 2) == 1 




```

```python
def distance_from_zero(x):
    return abs(subtract(0,x))

distance_from_zero(-25)
```
