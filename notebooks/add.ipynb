### Sum list


This function sums the elements in a list using the add function.

```python
def sum_list(values):
    assert type(values) == list or tuple, "Values should be a list or tuple"
    c = 0
    for i in values:
        c = add(c, i)
    return c


sum_list([1, 2, 3])
assert sum_list([1, 2, 3]) == 6
```
